# BRONZE_SIGTAP

**Origem:** `TB_SIGTAW.DBF` do pacote `TAB_SIA.zip` (`ftp://ftp.datasus.gov.br/dissemin/publicos/SIASUS/200801_/Auxiliar/`): tabela de procedimentos do SUS (SIGTAP).

Colunas mantidas com os nomes da origem (camada bronze = cópia fiel).

**Estratégia de carga:** full (`TRUNCATE` + insert direct-path em lotes).

```python
import sys
sys.path.append("..")
from dw_utils import baixar_tab_sia, carregar_dbf, executar, log_carga, consultar
import tempfile
```

```python
COLUNAS = ['IP_COD', 'IP_DSCR']

with tempfile.TemporaryDirectory() as pasta:
    arquivos = baixar_tab_sia(pasta)
    executar("TRUNCATE TABLE BRONZE_SIGTAP")
    lidos, gravados = carregar_dbf(arquivos["TB_SIGTAW.DBF"], "BRONZE_SIGTAP", COLUNAS)

log_carga("BRONZE_SIGTAP", lidos, gravados)
```

```python
consultar("SELECT * FROM BRONZE_SIGTAP FETCH FIRST 10 ROWS ONLY")
```

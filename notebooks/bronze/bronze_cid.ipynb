# BRONZE_CID

**Origem:** `S_CID.DBF` do pacote `TAB_SIA.zip` (`ftp://ftp.datasus.gov.br/dissemin/publicos/SIASUS/200801_/Auxiliar/`): Classificação Internacional de Doenças (CID-10): categorias e subcategorias.

Colunas mantidas com os nomes da origem (camada bronze = cópia fiel).

**Estratégia de carga:** full (`TRUNCATE` + insert direct-path em lotes).

```python
import sys
sys.path.append("..")
from dw_utils import baixar_tab_sia, carregar_dbf, executar, log_carga, consultar
import tempfile
```

```python
COLUNAS = ['CD_COD', 'CD_DESCR']

with tempfile.TemporaryDirectory() as pasta:
    arquivos = baixar_tab_sia(pasta)
    executar("TRUNCATE TABLE BRONZE_CID")
    lidos, gravados = carregar_dbf(arquivos["S_CID.DBF"], "BRONZE_CID", COLUNAS)

log_carga("BRONZE_CID", lidos, gravados)
```

```python
consultar("SELECT * FROM BRONZE_CID FETCH FIRST 10 ROWS ONLY")
```

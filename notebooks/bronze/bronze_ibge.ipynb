# BRONZE_IBGE

**Origem:** API de localidades do IBGE, visão "nivelada":
`https://servicodados.ibge.gov.br/api/v1/localidades/municipios?view=nivelado`

Carregamos todos os municípios do Brasil, e não só Alagoas, porque pacientes de outros
estados também são atendidos em Alagoas.

**Estratégia de carga:** full (`TRUNCATE` + `INSERT`).

```python
import sys
sys.path.append("..")
from dw_utils import substituir, consultar
import requests
```

```python
URL = "https://servicodados.ibge.gov.br/api/v1/localidades/municipios?view=nivelado"

resposta = requests.get(URL, timeout=120)
resposta.raise_for_status()

linhas = [(str(m["municipio-id"]), m["municipio-nome"], m["UF-sigla"]) for m in resposta.json()]
substituir("BRONZE_IBGE", ["COD_IBGE_7", "NOME_MUNICIPIO", "UF"], linhas)
```

```python
consultar("SELECT UF, COUNT(*) AS MUNICIPIOS FROM BRONZE_IBGE GROUP BY UF ORDER BY UF")
```

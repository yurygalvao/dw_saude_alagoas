# BRONZE_CNES

**Origem:** `CADGERAL.DBF` do pacote `TAB_SIA.zip` (`ftp://ftp.datasus.gov.br/dissemin/publicos/SIASUS/200801_/Auxiliar/`): cadastro dos estabelecimentos de saúde de **Alagoas** (CNES).

Colunas mantidas com os nomes da origem (camada bronze = cópia fiel).

**Estratégia de carga:** full (`TRUNCATE` + insert direct-path em lotes).

```python
import sys
sys.path.append("..")
from dw_utils import baixar_tab_sia, carregar_dbf, executar, log_carga, consultar
import tempfile
```

```python
COLUNAS = ['CNES', 'FANTASIA', 'RAZ_SOCI', 'CODUFMUN']

with tempfile.TemporaryDirectory() as pasta:
    arquivos = baixar_tab_sia(pasta)
    executar("TRUNCATE TABLE BRONZE_CNES")
    lidos, gravados = carregar_dbf(arquivos["CADGERAL.DBF"], "BRONZE_CNES", COLUNAS)

log_carga("BRONZE_CNES", lidos, gravados)
```

```python
consultar("SELECT * FROM BRONZE_CNES FETCH FIRST 10 ROWS ONLY")
```

# SILVER_ESTABELECIMENTO

Cadastro de estabelecimentos de Alagoas, com uma linha por CNES. O nome é o nome
fantasia, ou a razão social quando o fantasia está vazio.

**Origem:** `BRONZE_CNES`.

**Estratégia de carga:** full (`TRUNCATE` + `INSERT /*+ APPEND */`), idempotente.

```python
import sys
sys.path.append("..")
from dw_utils import recarregar, consultar
```

```python
SQL = """
INSERT /*+ APPEND */ INTO SILVER_ESTABELECIMENTO
    (COD_CNES, NOME_ESTABELECIMENTO, RAZAO_SOCIAL, COD_IBGE_6)
SELECT LPAD(TRIM(CNES), 7, '0'),
       MAX(UPPER(TRIM(NVL(FANTASIA, RAZ_SOCI)))),
       MAX(UPPER(TRIM(RAZ_SOCI))),
       MAX(TRIM(CODUFMUN))
FROM BRONZE_CNES
WHERE TRIM(CNES) IS NOT NULL
GROUP BY LPAD(TRIM(CNES), 7, '0')
"""
```

```python
recarregar("SILVER_ESTABELECIMENTO", SQL, origem="BRONZE_CNES")
```

```python
consultar("SELECT * FROM SILVER_ESTABELECIMENTO FETCH FIRST 10 ROWS ONLY")
```

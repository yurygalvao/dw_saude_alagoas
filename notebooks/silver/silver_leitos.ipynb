# SILVER_LEITOS

Leitos por estabelecimento e competência: soma dos tipos de leito de cada
estabelecimento no mês (existentes e disponíveis ao SUS).

**Origem:** `BRONZE_LEITOS`.

**Estratégia de carga:** full (`TRUNCATE` + `INSERT /*+ APPEND */`), idempotente.

```python
import sys
sys.path.append("..")
from dw_utils import recarregar, consultar
```

```python
SQL = """
INSERT /*+ APPEND */ INTO SILVER_LEITOS
    (DT_COMPETENCIA, COD_CNES, QTD_LEITOS_EXISTENTES, QTD_LEITOS_SUS)
SELECT TO_DATE(COMPETEN || '01', 'YYYYMMDD'),
       CNES,
       SUM(NVL(QT_EXIST, 0)),
       SUM(NVL(QT_SUS, 0))
FROM BRONZE_LEITOS
GROUP BY COMPETEN, CNES
"""
```

```python
recarregar("SILVER_LEITOS", SQL, origem="BRONZE_LEITOS")
```

```python
consultar("SELECT * FROM SILVER_LEITOS FETCH FIRST 10 ROWS ONLY")
```
